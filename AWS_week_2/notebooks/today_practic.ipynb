### Step 2: Import Modular Functions from `src`

```python
from src import (
    load_faqs,
    clean_text,
    chunk_text,
    prepare_faq_documents,
    initialize_chroma_collection,
    index_faqs_to_chroma,
    query_faqs,
    print_search_results,
    search_and_display,
)
print('✅ All modular functions successfully imported!')
```

### Step 3: Ingest FAQ Dataset

```python
df_faqs = load_faqs('data/faqs.csv')
print(f'Total FAQs loaded: {len(df_faqs)}')
df_faqs.head(3)
```

### Step 4: Chunk Text with Sliding Window

```python
documents, metadatas, ids = prepare_faq_documents(df_faqs, chunk_size=200, chunk_overlap=40)
print(f'Prepared {len(documents)} overlapping chunks for indexing.')
print('Sample Document:', documents[0])
```

### Step 5: Initialize ChromaDB Collection & Index Documents

```python
collection = initialize_chroma_collection('practice_faq_collection')
indexed_count = index_faqs_to_chroma(collection, documents, metadatas, ids)
print(f'Successfully indexed {indexed_count} chunks into collection!')
```

### Step 6: Query Semantic Search Engine

```python
queries = [
    'How can I deploy and train machine learning models on AWS?',
    'What is the timeout limit for serverless functions?',
    'Which GPU instance family is optimized for deep learning training?'
]

for q in queries:
    search_and_display(collection, q, top_k=3)
```
